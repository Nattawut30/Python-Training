```python
""" Chapter 05: Support Vector Machines"""

# SVM shine with small to mediam-sized nonlinear datasets (hundreds to thousands of instances), especially for classification tasks.
# The two classes can clearly be separated easily with a straight line (they are linear separable)
# You can think of an SVM classifier as fitting the widest possible street (represented by the parallel dashed lines) between the classes. This is called large margin classification.

# Adding more training instances "off the street" will not affect the decision boundary at all.
# These instances are called the support vectors
# THe widest possible street is close to horiontal after feature scaling (using Sickit-Learn's StandardScaler)
```

```python
""" Soft Margin Classification """

# If we strictly impose tha all instances must be off the street and on the correct side, this is called "hard margin classification"
# It only works if the data is linearly separable.
# it is sensitive to outliers.
# Without the outliers, and the model will probably not generalize as well.

# so, we need to use a more flexible model.
# The objective is to find a good balalnce between keeping the street as large as possible and limiting the margin violations.
# It called "Soft Margin Classification"

# When creating an SVM model using Scikit-Learn, You can specify several hyperparameters, including the regularization hyperparameter C.
# Reducing C results in more instances supporting the street, there's less rist of overfitting.
# But if reducing it too much, the model ends up underfitting.
# The model with C=100 looks like it will generalize better than the one with C=1.

```

```python
# Loads th iris dataset and trains a linear SVM classifier to detect "Iris virginca" flowers
# The pipeline first scales the features, then uses a LinearSVC with C=1

from sklearn.datasets import load_iris
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC

iris = load_iris(as_frame=True)
X = iris.data[["petal length (cm)", "petal width (cm)"]].values
y = (iris.target == 2) # iris virginica

svm_clf = make_pipeline(StandardScaler(), LinearSVC(C=1, random_state=42))

svm_clf.fit(X, y)
```

```python
# model predictions

X_new = [[5.5, 1.7], [5.0, 1.5]]
svm_clf.predict(X_new)
```

```python
# The first plant is classifid as an Iris virginica, the second is not
# Measure the signed distance between each instance and the decision boundary

svm_clf.decision_function(X_new)
```

```python
# LinearSVC does not have a predict_proba() method to estimate the class probabilities.
# This requires using 5-fold cross-validation to generate out-of-sample predictions for every instance in the training set,
# then training a LogisticRegression model, so it will slow down training considerably.
```

```python
""" Nonlinear SVM Classification """

# One approach to handling nonlinear datasets is to add more features, such as polynomial features.
# Create a pipeline containing a PolynomialFeatures transformer followed by a StandardScaler and LinearSVC classifier.

# Try the moon datasets using make_moons() function
# a toy dataset for binary classification in which the data points are shaped as two interleaving cresent moons

from sklearn.datasets import make_moons
from sklearn.preprocessing import PolynomialFeatures

X, y = make_moons(n_samples=100, noise=0.15, random_state=42)

polynomial_svm_clf = make_pipeline(
    PolynomialFeatures(degree=3),
    StandardScaler(),
    LinearSVC(C=10, max_iter=10_000, random_state=42)
)
polynomial_svm_clf.fit(X, y)
```

```python
""" Polynomial Kernel"""

# This can work great with all sorts of machine learning.
# This method cannot deal with very complex datasets.
# The model too slow.

# The kernel trick makes it possible to get the same result as if you had added many polynomial features.

from sklearn.svm import SVC

poly_kernel_svm_clf = make_pipeline(StandardScaler(),
                                    SVC(kernel="poly", degree=3, coef0=1, C=5))
poly_kernel_svm_clf.fit(X, y)
```

```python
# If your model is overfitting, you might want to reduce the polynomial degree.
# If it is underfitting, you can try increasing it. 
# The hyperparameter coef0 controls how much the model is influenced by high-degree terms versis low-degree terms.

# It's good to have a sense of what each hyperparameter actually does and how it may interact with other hyperparameters.
```

```python
""" Similarity Features """

# Measures how much each instance resembles a particular landmark
# THis is a bell-shaped function varying from 0 (very far away from the landmark) to 1 (at the landmark)
# The simples approach is to create a landmark at the location of each and every instance in the dataset.
# The downside is that a training set with "m" instances and "n" features gets transformed into a training set with "m" instances and "m" features
```

```python
" Gaussian RBF Kernel "

# Just like the polynomial features = useful with any machine learning algorithm but it maybe troble for large datasets and expensive
rbf_kernel_svm_clf = make_pipeline(StandardScaler(),
                                   SVC(kernel="rbf", gamma=5, C=0.0001))
rbf_kernel_svm_clf.fit(X, y)
```

```python
# Increasing gamme makes the bell-shaped curved narrower
# A small gamma value makes the bell-shaped curve wider
# A gamma acts like a regularization hyperparameter
# If your model is overfitting, you should reduce gamma,
# if it's underfitting you should increase gamma (similar to the C hyperparameter)

# "string kernel" are sometimes used when classifying text documents or DNA sequence.

# You should always try the linear kernel first.
# if the training set is very large = The LinearSVC class is much faster than SVC(kernel="linear")
# If the training set is not large = try kernelized SVMs, starting with the Gaussian RBF kernel.
```

```python
""" SVM Classes and Computational Complexity """

# The LinearSVC class is based on the liblinear library
# Its training time complexity is roughly O(m X n)
# The algorithm takes longer if you require very high precision.

# The SVC class is based on the libsvm library
# The training time complexity is usually between O(m^2 x n) and O(m^3 x n)
# This algorithm is best for small or medium-sized nonlinear training sets.
# The algirithm scales roughly with the average number of non-zero featres per instances

# The SGDClassifier class also performs large margin classification by default
# the learning_rate-can be adjusted to produce similar results as the linear SVMs.
# For training it uses stochastic gradient descent which allows incremental learning and uses little memory,
# So you can use it to train a model on a large dataset that does not fit in RAM. Complexity is O(m x n)
```

```python
""" SVM Regression """

# SVM Regression tries to fit as many instances as possible on the street while limiting margin violations.
# The width of the street is controlled by a hyperparameter, E

# Reducing e increases the number of support vectors, which regularizes the model.
# if you add more training instances within the margin, it will not affect the model's predictions; thus, the model is said to be e-insensitive.

# Use Scikit-Learn's LinearSVR class to perform linear SVM regression.
from sklearn.svm import LinearSVR

svm_reg = make_pipeline(StandardScaler(),
                        LinearSVR(epsilon=0.5, random_state=42))
svm_reg.fit(X, y)
```

```python
""" Under the Hood of Linear SVM Classifiers """

# If the result is positive, then the predicted class Y^ is the positive class 1.
# If it is the negative class 0, This is exactly like LogisticRegression

# Start with the width of the stree: To make it larger, we need to make 'w' smaller
# So, we need to keep 'w' as small as possible. note that the bias term 'b' has no influence on the size of the margin:
# tweaking it just shift the margin around, without affecting its size.

# We also want to avoid margin violations, we need the decision function to be greater than 1 for all positive training instances and lower than -1 for negative training instances.
# Optimization algorithm often work much better on differentiable functions.

# 2 conflict objective: 
# 1. Make the slack variables as small as possible to reduce the margin violations,
# 2. make 1/2 w^t w as small as possible ti increase the margin.

# That's where the C hyperparameter comes in
# It allows us to define the trade-off between these 2 objective problems.

# The hard margin and soft margin problems are both convex quadratic optimization problems with linear constraints.
# A.K.A "quadratic programming (QP) problems"

# Using a QP solver is one way to train an SVM.
# Another is to use gradient descent to minimize the hinge loss or the squared hinge loss

# The further away an instance is from the correct side of the margin, the higher the loss
# It grows linearly for the hinge loss and quadratically for the squared hinge loss. This makes the squared hinge loss more sensitive to outliers.

# LinearSVC uses the squared hinge loss
# SGDClassifier uses the hinge loss
# Both class allow you to pick a setting the loss hyperparameter to "hinge" or "sqaured_hinge"
# ** The SVC class's optimization algorithm finds a similar solution as minimizaing the hinge loss.
```

```python
""" The Dual Problem """

# The solution to the dual problem typically gives a lower bound to the solution of the primal problem but under some conditions it can have the same solution as the primal problem
# the SVM problem happens to meet the condition
# So you can choose to solve the primal problem or the dual problem. both will have the same solution.

# The dual problem is faster to solve than the primal one when the number of training instances is smaller than the number of features.
# The dual problem makes the kernel trick possible, whule the primal problem does not!
```

```python
""" Kernelized SVMs """

# what if you apply a second-degree polynomial transformation to a 2D training set likes... the moon training set
# then you train a linear SVM classifier on that transformed training set?
# Ans: The transformed vector is 3D, not 2D!

# What if you apply the second-degree polynomial mapping and then try to compute the dot product of the transformed vectors?
# Ans: The dot product of the transformed vectors is equal to the square of the dot product of the original vectors!

# Key insights:
# If you apply the transformation 0/ to all training instances, then the dual problem will contain the dot product
# but if 0/ is the second-degree polynomial transformation defined, then you can repalce this dot product of transformed vectors simply

# So... you don't need to transform the training instances at all!
# Just replace the dot product. by its square.

# The function K(a, b) = (a^T b)^2 is the second-degree polynomial kernel.
# In ML, A kernel is a function capable of computing the dot product based only on the original vectors a and b,
# without having to compute or even know about the transformation

# Common kernel:
# Linear = K(a, b) = a^T * b
# Polynomial = K(a, b) = (y * a^t + r)^d
# Gaussian RBF: K(a, b) = exp(-y || a - b)^2
# Sigmoid = K(a, b) = tanh(y * a^t * B + r)
```

```python
""" Mercer's Theorem """

# K must be continunous and symmetric in its arguments so that K(a, b) = K(b, a)
# You can use 'K' as a kernel because you know o/ exits, even if you don't know what o/ is
# So its's a good thing you don't need to actually perform mapping!
# Note that some frequently used kernels such as the sigmoid kernel don't respect all of Mercer's condition, yet they generally work well in practice.

# The good news is... you can plug the formula from w^ into the decision function for a new instance x^(n),
# and you get an equation with only dot products between input vectors.
# This make it possible to use the kernel trick
# and also, Making prediction involves computing the dot product of the new input vector x^(n) with only the support vectors,
# not all the training instances.

# A little headache, but that's the point and perfectly normal.
# It's an unfortunate side effect of the kernel trick!

# The End.
```
